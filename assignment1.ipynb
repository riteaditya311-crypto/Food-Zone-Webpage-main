```python
import pandas as pd
import numpy as np

np_array = np.array([10, 20, 30, 40, 50])
pd_series = pd.Series(np_array)

print("NumPy Array:\n", np_array)
print("\nPandas Series:\n", pd_series)
```

```python
date_series = pd.Series(['2023-01-01', '2024-05-15', '2025-10-31'])
datetime_series = pd.to_datetime(date_series)

df_dates = pd.DataFrame({
    'Date': datetime_series,
    'Day': datetime_series.dt.day,
    'Week Number': datetime_series.dt.isocalendar().week,
    'Day of Year': datetime_series.dt.dayofyear,
    'Day of Week': datetime_series.dt.weekday
})

display(df_dates)
```

```python
word_series = pd.Series(['hello', 'world', 'pandas', 'numpy'])

def capitalize_first_last(val):
    if len(val) <= 1:
        return val.upper()
    return val[0].upper() + val[1:-1] + val[-1].upper()

capitalized_series = word_series.apply(capitalize_first_last)
print("Original Series:\n", word_series)
print("\nCapitalized Series:\n", capitalized_series)
```

```python
data = pd.Series([12, 45, 2, 89, 34, 55, 23, 77, 10])

print("Minimum:", data.min())
print("Maximum:", data.max())
print("Median:", data.median())
print("25th Percentile:", data.quantile(0.25))
print("75th Percentile:", data.quantile(0.75))
```

```python
print("Series Mean:", data.mean())
print("Series Standard Deviation:", data.std())
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([5, 4, 3, 2, 1])

distance = np.linalg.norm(s1 - s2)
print("Series 1:", s1.tolist())
print("Series 2:", s2.tolist())
print("Euclidean Distance:", distance)
```
